# Missile Trajectory Analysis & Visualization

This notebook loads the simulated missile trajectory datasets and demonstrates:
1. Trajectory profile (Altitude vs. Range)
2. Flight parameters vs. Time (Speed, Acceleration, Dynamic Pressure, Mach Number)
3. Kalman Filter smoothing for radar simulation

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
from src.simulation import run_simulation
from src.kalman import apply_kalman_filter
```

## 1. Run Simulation or Load Data

```python
# Load QRSAM data or run live simulation
qrsam_df = run_simulation('QRSAM', dt=0.5)
print(f"Generated {len(qrsam_df)} time steps")
qrsam_df.head()
```

## 2. Flight Trajectory (Altitude vs Range)

```python
plt.figure(figsize=(10, 5))
plt.plot(qrsam_df['Range_km'], qrsam_df['Altitude_km'], label='QRSAM Trajectory', color='royalblue', lw=2)
plt.title('Missile Flight Profile: Altitude vs Range')
plt.xlabel('Range (km)')
plt.ylabel('Altitude (km)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

## 3. Speed & Mach Number vs Time

```python
fig, ax1 = plt.subplots(figsize=(10, 5))

color = 'tab:blue'
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Speed (m/s)', color=color)
ax1.plot(qrsam_df['Time_s'], qrsam_df['Speed_ms'], color=color, lw=2, label='Speed (m/s)')
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Mach Number', color=color)
ax2.plot(qrsam_df['Time_s'], qrsam_df['Mach_Number'], color=color, lw=2, linestyle='--', label='Mach Number')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('QRSAM Speed and Mach Number Evolution')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 4. Radar Synthetic Noise & Kalman Filter Smoothing

```python
smoothed_df = apply_kalman_filter(qrsam_df, process_noise=2.0, measurement_noise=50.0)

plt.figure(figsize=(10, 6))
plt.plot(smoothed_df['Time_s'], smoothed_df['Altitude_m'], label='True Trajectory', color='green', lw=2)
plt.scatter(smoothed_df['Time_s'][::5], smoothed_df['Altitude_Measured'][::5], label='Radar Measurement (Noisy)', color='crimson', s=15, alpha=0.6)
plt.plot(smoothed_df['Time_s'], smoothed_df['Altitude_Smoothed'], label='Kalman Filter Track', color='navy', linestyle='--', lw=2)
plt.title('Radar Tracking & Kalman Filter Smoothing')
plt.xlabel('Time (s)')
plt.ylabel('Altitude (m)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
